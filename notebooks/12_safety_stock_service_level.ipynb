# 12 — Safety stock and service level
Study the trade-off between service level and inventory tied up in safety stock.

```python
import pandas as pd
from demand_supply.forecasting.inventory import safety_stock
rows=[]
for service in [.80,.90,.95,.975,.99]:
    rows.append({'service_level':service,'safety_stock':safety_stock(daily_std=8,lead_time_days=4,service_level=service)})
pd.DataFrame(rows)
```

Higher service levels reduce stockout risk but increase working capital and potential waste. Perishable and non-perishable categories should not share one service target blindly.
