# 18 — Real-time POS simulation
Simulate sale events that can be posted to the FastAPI endpoint or n8n webhook to test the live path.

```python
from datetime import datetime
import uuid, random, json
products=[('MILK-1L','dairy',1.29),('BREAD-WHEAT','bakery',2.49),('BANANA-KG','produce',1.89)]
def event():
    p,c,price=random.choice(products)
    return {'event_id':str(uuid.uuid4()),'timestamp':datetime.now().astimezone().isoformat(),'store_id':'GOE-001','product_id':p,'category':c,'quantity':random.randint(1,4),'unit_price':price,'channel':'store'}
event()
```
