# 03 — Data quality and leakage
Audit missing days, duplicates, impossible quantities, price jumps and future-information leakage.

```python
import pandas as pd
df = pd.read_parquet('../data/processed/demo_daily_sales.parquet')
df['ds'] = pd.to_datetime(df['ds'])
```

```python
checks = {
 'duplicate_keys': df.duplicated(['ds','store_id','product_id']).sum(),
 'negative_units': (df.y < 0).sum(),
 'missing_target': df.y.isna().sum(),
 'missing_price': df.unit_price.isna().sum(),
}
checks
```

```python
coverage = (df.groupby('product_id').ds.agg(['min','max','nunique']).sort_values('nunique'))
coverage
```

## Leakage rule
Only use variables known by forecast creation time. Realized future weather, future sales, and post-event footfall are forbidden; use forecast weather and pre-published calendars instead.

```python
assert checks['negative_units'] == 0
assert checks['missing_target'] == 0
```
