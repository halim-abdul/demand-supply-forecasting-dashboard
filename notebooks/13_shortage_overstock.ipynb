# 13 — Shortage and overstock classification
Create operational labels from inventory position, reorder point and days of cover.

```python
import pandas as pd
from demand_supply.forecasting.inventory import inventory_plan
fc = pd.DataFrame({'ds':pd.date_range('2026-10-07', periods=28),'yhat':[20]*28})
plans=[]
for stock in [10,30,60,100,200,400]:
    p=inventory_plan(fc,on_hand=stock,lead_time_days=3,daily_std=4); p['on_hand']=stock; plans.append(p)
pd.DataFrame(plans)
```

Prioritize actions by economic impact, not status alone: stockout cost, margin, perishability and supplier lead time can turn the same numerical risk into different decisions.

```python
risk = pd.DataFrame(plans).sort_values(['shortage_risk','recommended_order_qty'], ascending=False)
risk[['on_hand','status','shortage_risk','days_cover','recommended_order_qty']]
```
