# 04 — Calendar and Lower Saxony holidays
Build deterministic date features and official public-holiday indicators for Niedersachsen.

```python
import pandas as pd
from demand_supply.features import calendar_features
from demand_supply.external.holidays_de import lower_saxony_holidays
```

```python
dates = pd.Series(pd.date_range('2026-01-01','2026-12-31',freq='D'))
cal = calendar_features(dates)
hol = lower_saxony_holidays('2026-01-01','2026-12-31')
hol
```

```python
features = cal.merge(hol, on='ds', how='left')
features['is_public_holiday'] = features.is_public_holiday.fillna(0)
features.query('is_public_holiday == 1').head(20)
```

Public holidays can produce both pre-holiday stock-up and closure-day zeros. Model them with historical evidence rather than assuming every holiday increases demand.
