# 17 — Sales anomaly detection
Flag unusual spikes/drops before retraining so data issues and one-off events are reviewed rather than silently learned.

```python
import pandas as pd
from demand_supply.forecasting.anomaly import robust_sales_anomalies
df=pd.read_parquet('../data/processed/demo_daily_sales.parquet')
sku=df.query("product_id == 'MILK-1L'")[['ds','y']].sort_values('ds')
flagged=robust_sales_anomalies(sku)
flagged.query('is_anomaly').tail(20)
```

```python
flagged.anomaly_score.describe()
```

Do not automatically delete anomalies: a festival spike or stockout dip may be real and operationally important. Attach cause labels when known.
