# 02 — POS exploratory analysis
Explore daily demand, revenue, weekday patterns and SKU concentration using demo or retailer-provided data.

```python
import pandas as pd
import plotly.express as px
df = pd.read_parquet('../data/processed/demo_daily_sales.parquet')
df['ds'] = pd.to_datetime(df['ds'])
df.head()
```

```python
summary = df.groupby('product_id').agg(units=('y','sum'), avg_daily=('y','mean'), volatility=('y','std')).sort_values('units', ascending=False)
summary
```

```python
daily = df.groupby('ds', as_index=False).y.sum()
px.line(daily, x='ds', y='y', title='Total daily units')
```

```python
weekday = df.assign(dow=df.ds.dt.day_name()).groupby('dow', as_index=False).y.mean()
weekday
```
