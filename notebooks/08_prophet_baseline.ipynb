# 08 — Prophet baseline
Fit one SKU, inspect trend/weekly seasonality and compare against a seven-day seasonal naive forecast.

```python
import pandas as pd
from demand_supply.forecasting.prophet_model import ProphetDemandModel
from demand_supply.forecasting.baselines import seasonal_naive
df = pd.read_parquet('../data/processed/demo_daily_sales.parquet')
sku = df.query("product_id == 'MILK-1L'")[['ds','y']].sort_values('ds')
```

```python
train, test = sku.iloc[:-56], sku.iloc[-56:]
model = ProphetDemandModel(regressors=[]).fit(train)
pred = model.predict(test[['ds']])
pred.head()
```

```python
naive = seasonal_naive(train, 56)
from demand_supply.forecasting.metrics import wape
{'prophet_wape': wape(test.y, pred.yhat), 'seasonal_naive_wape': wape(test.y, naive.yhat)}
```

Do not deploy Prophet merely because it is sophisticated: it should consistently beat the seasonal naive baseline on rolling backtests.
