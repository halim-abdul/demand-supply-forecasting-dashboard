# 10 — Rolling-origin backtesting
Evaluate forecasts the way the system will operate in time: train only on the past, predict a future horizon, roll forward, repeat.

```python
import pandas as pd
from demand_supply.forecasting.backtest import rolling_backtest
df = pd.read_parquet('../data/processed/demo_daily_sales.parquet')
sku = df.query("product_id == 'BREAD-WHEAT'")[['ds','y']].sort_values('ds')
scores = rolling_backtest(sku, horizon=28, folds=4)
scores
```

```python
scores.mean(numeric_only=True)
```
