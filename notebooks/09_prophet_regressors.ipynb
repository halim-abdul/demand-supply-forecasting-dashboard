# 09 — Prophet with external regressors
Add holidays, events, promotions, weather and catchment variables, then quantify whether they improve out-of-sample error.

```python
import pandas as pd
from demand_supply.features import enrich_daily_sales
from demand_supply.forecasting.prophet_model import ProphetDemandModel
df = pd.read_parquet('../data/processed/demo_daily_sales.parquet')
sku = df.query("product_id == 'ICECREAM'").copy()
enriched = enrich_daily_sales(sku)
enriched.columns.tolist()
```

```python
numeric = ['event_lift','is_public_holiday','promotion','population_share']
for c in numeric:
    if c in enriched: enriched[c] = enriched[c].fillna(0).astype(float)
model = ProphetDemandModel(regressors=numeric).fit(enriched.iloc[:-56])
pred = model.predict(enriched.iloc[-56:])
pred[['ds','yhat','yhat_lower','yhat_upper']].head()
```

A regressor is useful only if it is available at prediction time and improves rolling validation. Remove unstable or redundant variables.
