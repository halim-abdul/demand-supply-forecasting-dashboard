# 16 — Forecast ensembles
Blend models by inverse validation error so a single model failure does not dominate decisions.

```python
import pandas as pd
from demand_supply.forecasting.ensemble import inverse_error_weights, blend_forecasts
errors={'prophet':0.18,'seasonal_naive':0.22,'xgb':0.16}
weights=inverse_error_weights(errors); weights
```

```python
ds=pd.date_range('2026-10-07',periods=7)
forecasts={
 'prophet':pd.DataFrame({'ds':ds,'yhat':[30,31,32,35,38,33,28]}),
 'seasonal_naive':pd.DataFrame({'ds':ds,'yhat':[29,30,31,34,37,32,27]}),
 'xgb':pd.DataFrame({'ds':ds,'yhat':[31,31,33,36,39,34,29]})}
blend_forecasts(forecasts,weights)
```

Weights should be recalculated from recent rolling validation and constrained to avoid unstable jumps.
