# 19 — Model monitoring
Track error, bias, interval coverage and business KPIs after deployment.

```python
import pandas as pd
from demand_supply.forecasting.metrics import mae,rmse,wape,bias
monitor=pd.DataFrame({'actual':[30,35,32,50,42,28,26],'forecast':[31,33,34,44,45,30,27],'lower':[24,26,26,35,36,22,21],'upper':[38,40,42,55,54,38,34]})
metrics={'mae':mae(monitor.actual,monitor.forecast),'rmse':rmse(monitor.actual,monitor.forecast),'wape':wape(monitor.actual,monitor.forecast),'bias':bias(monitor.actual,monitor.forecast),'coverage':((monitor.actual>=monitor.lower)&(monitor.actual<=monitor.upper)).mean()}
metrics
```

Alert on sustained bias, WAPE degradation, interval under-coverage, stockout-rate increases and data freshness failures. Retraining should be evidence-driven, not scheduled blindly.

```python
thresholds={'wape':0.25,'abs_bias':0.10,'coverage_min':0.80}
{'wape_alert':metrics['wape']>thresholds['wape'],'bias_alert':abs(metrics['bias'])>thresholds['abs_bias'],'coverage_alert':metrics['coverage']<thresholds['coverage_min']}
```
