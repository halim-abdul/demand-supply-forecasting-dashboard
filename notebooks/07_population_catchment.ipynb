# 07 — Population and catchment
Use district population as a contextual feature for store catchment, never as a direct substitute for observed demand.

```python
import pandas as pd
pop = pd.read_csv('../data/reference/goettingen_population_by_district.csv')
pop['share'] = pop.population / pop.population.sum()
pop.sort_values('population', ascending=False)
```

```python
pop.population.sum(), pop.share.sum()
```

A practical extension is travel-time weighted catchment: estimate each district's probability of choosing a store from distance, transport links and competitor locations, then multiply by district population.

```python
pop.assign(relative_index=pop.population/pop.population.mean())[['district','population','share','relative_index']].head()
```
